```python
# Homework 7
# Exercises 16.1 and 16.4

import tensorflow as tf
import time

from tensorflow.keras.datasets import mnist, fashion_mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
```

```python
# Exercise 16.1

# Load MNIST
(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train = x_train.reshape(60000, 28, 28, 1).astype('float32') / 255
x_test = x_test.reshape(10000, 28, 28, 1).astype('float32') / 255

# MNIST model
mnist_model = Sequential([
    Conv2D(64, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

mnist_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

start = time.time()

mnist_model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=64
)

mnist_time = time.time() - start

mnist_loss, mnist_accuracy = mnist_model.evaluate(x_test, y_test)

print("MNIST accuracy:", mnist_accuracy)
print("MNIST training time:", mnist_time, "seconds")
```

```python
# Load Fashion-MNIST
(fashion_train, fashion_y_train), (fashion_test, fashion_y_test) = fashion_mnist.load_data()

fashion_train = fashion_train.reshape(60000, 28, 28, 1).astype('float32') / 255
fashion_test = fashion_test.reshape(10000, 28, 28, 1).astype('float32') / 255

# Same model using Fashion-MNIST
fashion_model = Sequential([
    Conv2D(64, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

fashion_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

start = time.time()

fashion_model.fit(
    fashion_train,
    fashion_y_train,
    epochs=5,
    batch_size=64
)

fashion_time = time.time() - start

fashion_loss, fashion_accuracy = fashion_model.evaluate(
    fashion_test,
    fashion_y_test
)

print("Fashion-MNIST accuracy:", fashion_accuracy)
print("Fashion-MNIST training time:", fashion_time, "seconds")
```

Exercise 16.1 Answer: The model performed better on MNIST with 99.15% accuracy compared to 90.87% on Fashion-MNIST. The training times were pretty similar. MNIST took about 668 seconds and Fashion-MNIST took about 652 seconds.

```python
# Exercise 16.4
# Remove the first Dense layer

removed_model = Sequential([
    Conv2D(64, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(10, activation='softmax')
])

removed_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

start = time.time()

removed_model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=64
)

removed_time = time.time() - start

removed_loss, removed_accuracy = removed_model.evaluate(x_test, y_test)

print("Accuracy with Dense layer removed:", removed_accuracy)
print("Training time:", removed_time, "seconds")
```

```python
# Add a Dense layer with 4096 neurons

large_model = Sequential([
    Conv2D(64, (3, 3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(4096, activation='relu'),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

large_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

start = time.time()

large_model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=64
)

large_time = time.time() - start

large_loss, large_accuracy = large_model.evaluate(x_test, y_test)

print("Accuracy with 4096 neuron layer:", large_accuracy)
print("Training time:", large_time, "seconds")
```

Exercise 16.4 Answer: Removing the first Dense layer barely changed the accuracy. It went from 99.15% to 99.06%, and the training time was about the same. Adding the 4096-neuron Dense layer lowered the accuracy to 98.69% and made training much slower. It took about 2,127 seconds compared to 668 seconds for the original model.
