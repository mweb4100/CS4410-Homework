```python
from sklearn.datasets import load_iris
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score, GridSearchCV
import matplotlib.pyplot as plt
```

```python
iris = load_iris()

X = iris.data
y = iris.target
```

```python
k_range = range(1, 31)
accuracy_scores = []

for k in k_range:
    knn = KNeighborsClassifier(n_neighbors=k)
    scores = cross_val_score(knn, X, y, cv=10)
    accuracy_scores.append(scores.mean())
```

```python
best_accuracy = max(accuracy_scores)
best_k = accuracy_scores.index(best_accuracy) + 1

print("Best k:", best_k)
print("Best accuracy:", best_accuracy)
```

```python
plt.plot(k_range, accuracy_scores)

plt.xlabel("Value of K for KNN")
plt.ylabel("Cross-Validated Accuracy")
plt.title("K Value vs. Cross-Validated Accuracy")

plt.show()
```

```python
plt.plot(k_range, accuracy_scores)

plt.xlim(11, 15)
plt.ylim(0.966, 0.981)

plt.xlabel("Value of K for KNN")
plt.ylabel("Cross-Validated Accuracy")
plt.title("K Value vs. Cross-Validated Accuracy")

plt.show()
```

```python
param_grid = {"n_neighbors": range(1, 31)}

grid = GridSearchCV(KNeighborsClassifier(), param_grid, cv=10)
grid.fit(X, y)

print("Best k:", grid.best_params_["n_neighbors"])
print("Best accuracy:", grid.best_score_)
```

The optimal k value is 13. Using 10-fold cross-validation, the KNN classifier achieved an accuracy of about 98%. GridSearchCV also confirmed that k = 13 is an optimal value.
